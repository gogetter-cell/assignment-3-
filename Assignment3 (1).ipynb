```python
# 1. Write a python function that takes a list of numbers as input and returns the sum of all the even numbers in the list.
def sum_even_numbers(numbers):
    sum = 0
    for num in numbers:
        if num % 2 == 0:
            sum += num
    return sum
```

```python
#2.Write a python function that accepts a string and returns the reverse of the string.
def reverse_string(string):
    return string[::-1]
```

```python
#3.Implement a python function that takes a list of integers and returns a new list containing squares of each number.
def square_numbers(numbers):
    squares = []
    for num in numbers:
        squares.append(num**2)
    return squares
```

```python
#4.write a python function to check whether the given number is prime or not from 1 to 200.
def is_prime(number):
    if number < 2:
        return "Neither prime nor composite number"
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return "Not a prime number"
    return "Prime Number"
```

```python
#5.Create a iterator class in python that generates a fibonaci sequence up to a specified number of terms.
def fibonacci_sequence(n):
    a, b = 0, 1
    for i in range(n):
        yield a
        a, b = b, a + b
```

```python
#6. Write a generator function in python that yields powers of 2 upto a given exponent.
def power_of_two(n):
    for i in range(n+1):
        yield 2 ** i
```

```python
#7.Implement a generator function that reads file line by line and yields each line as a string.
def read_file(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line
```

```python
#8.Use a lambda function to sort a list of tuples based on second element of each tuple.
def sort_list_of_tuples(list_of_tuples):
    return sorted(list_of_tuples, key=lambda x: x[1])
```
